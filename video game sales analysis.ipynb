```python
i = 'Global_Sales'
```

```python
q1 = np.percentile(df[i],25)
q3 = np.percentile(df[i],75)

iqr = q3-q1

print(q1)
print(q3)
print(iqr)
```

```python
c1 = q1-1.5*(iqr)
c2 = q3+1.5*(iqr)

t = df[(df[i]>c2) | (df[i]<c1)]
t=t.index
df=df.drop(labels=t)
t
```

```python
#Publisher Performance Analysis
```

```python
top_performing = df.groupby('Publisher')['Global_Sales'].sum().nlargest(10)
top_performing.reset_index()
```

```python
plt.figure(figsize=(12,6))
sb.barplot(x=top_performing.index,y=top_performing.values)
plt.xticks(rotation=90)
plt.xlable("Publisher")
plt.ylabel("Global Sales")
plt.title('Top 10 Publishers by Global Sales')
plt.tight_layout()
plt.show()
```
